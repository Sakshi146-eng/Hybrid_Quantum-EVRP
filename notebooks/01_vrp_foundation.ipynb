# Step 1: Classical VRP Foundation & Distance Calculation

This notebook demonstrates the classical Vehicle Routing Problem (VRP) data representation, Euclidean distance matrix calculation, route distance/cost evaluation, and classical constraint validation.

> **Note:** This stage establishes the classical baseline. No QUBO, QAOA, or Reinforcement Learning formulation is included in this notebook.

```python
import sys
from pathlib import Path
import numpy as np

# Add root project directory to sys.path
root_dir = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

from src.data.loader import DataLoader
from src.vrp.instance import VRPInstance
from src.vrp.distance import compute_distance_matrix, validate_distance_matrix
```

## 1. Load Toy VRP Instance (3 Customers + Depot)

We load a small, deterministic VRP instance with 3 customer nodes and 1 depot node with fixed integer coordinates.

```python
loader = DataLoader()
instance = loader.load_vrp_instance(instance_type="deterministic", num_customers=3)

print(f"Total Nodes : {instance.num_nodes}")
print(f"Depot Index : {instance.depot_idx}")
print(f"Capacity    : {instance.vehicle_capacity}")
print("\nCoordinates:")
for i, c in enumerate(instance.coordinates):
    node_name = 'Depot' if i == instance.depot_idx else f'Customer {i}'
    print(f"  Node {i} ({node_name}): ({c[0]:.1f}, {c[1]:.1f}) | Demand: {instance.demands[i]:.1f}")
```

## 2. Distance Matrix Calculation & Symmetry Check

```python
print("Pairwise Distance Matrix:")
print(np.round(instance.distance_matrix, 2))

is_symmetric = validate_distance_matrix(instance.distance_matrix)
print(f"\nDistance Matrix Valid & Symmetric: {is_symmetric}")
```

## 3. Candidate Route Evaluation & Validation

```python
# Candidate Route: Depot (0) -> Cust 1 -> Cust 3 -> Cust 2 -> Depot (0)
route = [0, 1, 3, 2, 0]
dist = instance.calculate_route_distance(route)
validation = instance.validate_route(route)

print(f"Candidate Route : {route}")
print(f"Total Distance  : {dist:.2f}")
print(f"Validation      : {'VALID' if validation['is_valid'] else 'INVALID'}")
print(f"Total Demand    : {validation['total_demand']} / {instance.vehicle_capacity}")
```
